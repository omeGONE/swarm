```python
import numpy as np
from pso import PSO
```

```python
def fitness(X: np.ndarray) -> np.ndarray:
    x = X[:, 0]
    y = X[:, 1]
    return -np.abs(np.sin(x) * np.cos(y) * np.exp(
        np.abs(1 - np.sqrt(x ** 2 + y ** 2) / np.pi)))
```

```python
bounds = np.array([
    [-10.0, 10.0],
    [-10.0, 10.0],
])
```

```python
pso = PSO(
    n_particles=30,
    topology="ring",
    n_neighbors=4,
    max_iter=200,
    random_state=42,
)
```

```python
import time
```

```python
s = time.perf_counter()
result = pso.minimize(func=fitness, bounds=bounds)
result.best_position_, result.best_value_, time.perf_counter() - s
```

```python
pso = PSO(
    n_particles=30,
    topology="star",
    max_iter=200,
    random_state=42,
)

s = time.perf_counter()
result = pso.minimize(func=fitness, bounds=bounds)
result.best_position_, result.best_value_, time.perf_counter() - s
```
